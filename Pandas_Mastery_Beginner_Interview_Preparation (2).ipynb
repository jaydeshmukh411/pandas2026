# Series

A Series is a one-dimensional labelled array capable of holding heterogeneous data.

## Components
- Data
- Index
- Data Type

## Facts
- Mutable values
- Immutable size
- Supports vectorized operations

```python
# Create Series from list
s1 = pd.Series([10,20,30], index=['a','b','c'])
print(s1)

# Create Series from dictionary
s2 = pd.Series({'Math':90,'Science':95,'English':88})
print(s2)
```

# DataFrame

A DataFrame is a two-dimensional labelled data structure.

## Properties
1. Heterogeneous Data
2. Size Mutable
3. Labelled Axes
4. Supports Missing Values
5. Powerful Indexing

```python
# List of lists
df1 = pd.DataFrame(
    [['John',25],['Mary',30]],
    columns=['Name','Age']
)
print(df1)

# List of dictionaries
df2 = pd.DataFrame([
    {'Name':'Alice','Age':28},
    {'Name':'Bob','Age':35}
])
print(df2)

# Dictionary of Series
df3 = pd.DataFrame({
    'A': pd.Series([1,2,3]),
    'B': pd.Series([4,5,6])
})
print(df3)
```

# Adding Columns

Methods:
- Assign scalar value
- Assign list
- Assign Series

```python
df = pd.DataFrame({
    'Name':['John','Mary','Alice'],
    'Age':[25,30,28]
})

# Scalar
df['Country'] = 'India'

# List
df['Score'] = [80,90,85]

# Series
bonus = pd.Series([5,10,15])
df['Bonus'] = bonus

print(df)
```

# Rename Columns

rename(columns={}, inplace=False)

Parameters:
- columns
- inplace

```python
# Rename single column
df.rename(columns={'Age':'Employee_Age'}, inplace=True)

# Rename multiple columns
df.rename(columns={
    'Score':'Exam_Score',
    'Bonus':'Extra_Bonus'
}, inplace=True)

print(df)
```

# Indexing and Data Selection

## Types
- loc (label based)
- iloc (position based)

```python
employees = pd.DataFrame({
    'Name':['John','Mary','Alice','Bob'],
    'Department':['IT','HR','IT','Finance'],
    'Salary':[50000,60000,55000,70000]
}, index=['E1','E2','E3','E4'])

print(employees)
```

## loc Examples

```python
# Access row
print(employees.loc['E1'])

# Access column
print(employees.loc[:, 'Name'])

# Multiple rows and columns
print(employees.loc[['E1','E3'], ['Name','Salary']])

# Conditional indexing
print(employees.loc[employees['Salary'] > 55000])
```

## iloc Examples

```python
# Single row
print(employees.iloc[0])

# Single column
print(employees.iloc[:,0])

# Subset
print(employees.iloc[0:3, 0:2])
```

# Interview Questions

1. Difference between Series and DataFrame?

* ANS =>
A Series is a one-dimensional Pandas object used for a single column of data, while a DataFrame is a two-dimensional tabular structure with rows and columns that can store multiple data types.

2. loc vs iloc?

* ANS=> loc is label-based indexing and is used with column names or conditions, whereas iloc is position-based indexing and works with numerical indexes.

3. Why is Pandas built on NumPy?
* ANS=> Pandas is built on NumPy because NumPy provides fast array operations, optimized memory usage, and efficient numerical computation.

4. What is heterogeneous data?
* ANS=> Heterogeneous data means data containing different data types together, such as integers, strings, floats, and boolean values in the same dataset.

5. How do you rename columns?
* ANS=> Columns can be renamed using the rename() function, usually with the columns parameter to map old names to new names.

6. How do you add columns?
* ANS=> New columns can be added by assigning values, lists, or Series objects using df['ColumnName'] = values.

7. How does Pandas handle missing values?
* ANS=> Pandas handles missing values using NaN and provides functions like fillna(), dropna(), and isnull() for managing them.

8. Explain DataFrame attributes.
* ANS=> DataFrame attributes provide information about the dataset, such as shape for dimensions, columns for column names, dtypes for data types, and index for row labels.

9. Difference between list of dictionaries and dictionary of Series?
* ANS=> A list of dictionaries is generally used for row-wise data creation, while a dictionary of Series is mainly used for column-wise structured data.

10. When would you choose loc over iloc?
* ANS=> I would choose loc when working with column names, labels, or conditional filtering because it makes the code more readable and easier to maintain.
