Scenario
A retail company has recorded weekly sales (in ₹ thousands) for 4 stores over 12 weeks. The data arrived as a flat 1D array of 48 values with some entries corrupted as np.nan. Your task is to restructure and analyze it.
Tasks
Reshape the 1D array into a (4, 12) matrix — rows = stores, columns = weeks.
Replace all NaN values with the column (weekly) mean across non-NaN stores.
Find the top-performing store (highest total annual sales).
Flag weeks where ALL stores sold below ₹50k (boolean mask).
Compute cumulative sales per store across the 12 weeks.

```python
import numpy as np

arr = np.random.randint(1000, 2000, 48).astype(float)

arr = arr.reshape(4, 12)

print("Original Array:")
print(arr)
arr[0, 2] = np.nan
arr[2, 5] = np.nan

# Replace NaN with column mean
col_mean = np.nanmean(arr, axis=0)

rows, cols = np.where(np.isnan(arr))

arr[rows, cols] = col_mean[cols]

print("\nSales Matrix After Replacing NaN:")
print(arr)

# Top-performing store
store_sales = np.sum(arr, axis=1)

top_store = np.argmax(store_sales)

print("\nTop Performing Store:")
print(f"Store {top_store + 1}")

print("\nStore Sales:")
print(store_sales)

# Flag low-sales weeks
low_sales_weeks = np.all(arr < 50, axis=0)

print("\nWeeks Where All Stores Sold Below 50:")
print(low_sales_weeks)

# Cumulative sales per store
cumulative_sales = np.cumsum(arr, axis=1)

print("\nCumulative Sales Per Store:")
print(cumulative_sales)
```
